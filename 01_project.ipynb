```python
# seperating x and y
```

```python
x = df.drop("price" , axis =1)
y = df["price"]
print(x.head())
print(y.head())
```

```python
from sklearn.model_selection import train_test_split

```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
model.intercept_
```

```python
cofficients = pd.DataFrame({
    "Feature": x.columns,
    "Cofficient": model.coef_
})
cofficients
```

```python
from sklearn.metrics import r2_score, mean_absolute_error , mean_squared_error
r2 =  r2_score(y_test,y_pred)
mae = mean_absolute_error(y_test,y_pred)
mse = mean_squared_error(y_test,y_pred)
rmse = np.sqrt(mse)

print("r2 :", r2)
print("MAE :", mae)
print("MSE :", mse)
print("RMSE :", rmse)




```

```python
residuals = y_test - y_pred

plt.scatter(y_test,y_pred)
plt.xlabel("Actual Price")
plt.ylabel("Predcted Price")
plt.title("Actual VS Predicted House Price")
```

```python


plt.scatter(y_pred,residuals)


plt.axhline(y=0)

plt.xlabel("Predcted Price")
plt.ylabel("Residuals")
plt.title("Residual Analysis")
```

```python
# Feature Engineering
```

```python
df ["rooms"] = df["bedrooms"] + df["bathrooms"]
```

```python
df["area_per_bedroom"] = (
    df["area_sqft"] / df["bedrooms"]
)
```

```python
df["location_access_score"] = (
    df["nearby_schools"] + df["location_score"]
)
```

```python
df.head()
```

```python
x_fe = df.drop("price", axis = 1)
y_fe = df["price"]
print(x_fe.head())
print(y_fe.head(5))
```

```python
# Feature Engineered Linear Regression
```

```python
x_trian_fe , x_test_fe , y_train_fe , y_test_fe = train_test_split(x_fe , y_fe , test_size=0.2)
```

```python
fe_model = LinearRegression()
fe_model.fit(x_trian_fe , y_train_fe)
fe_pred = fe_model.predict(x_test_fe)
```

```python
fe_mae = mean_absolute_error(y_test_fe , fe_pred)
fe_mse = mean_squared_error(y_test_fe , fe_pred)
fe_rmse =np.sqrt(fe_mse)
fe_r2 = r2_score(y_test_fe , fe_pred)


print("Feature Engineered Linear Regression")
print("MAE:", fe_mae)
print("MSE:", fe_mse)
print("RMSE:", fe_rmse)
print("R²:", fe_r2)
```

```python
from sklearn.linear_model import Ridge
ridge_model = Ridge(alpha=1.0)
ridge_model.fit(x_trian_fe , y_train_fe)
ridge_pred = ridge_model.predict(x_test_fe)
```

```python
ridge_mae = mean_absolute_error(y_test_fe , ridge_pred)
ridge_mse = mean_squared_error(y_test_fe , ridge_pred)
ridge_rmse =np.sqrt(ridge_mse)
ridge_r2 = r2_score(y_test_fe , ridge_pred)


print("Ridge Regression")
print("MAE:", ridge_mae)
print("MSE:", ridge_mse)
print("RMSE:", ridge_rmse)
print("R²:", ridge_r2)
```

```python
comparison = ({
    "model":[
    "Multiple Linear Regression",
    "Feature Engninerred Linear Regression",
    "Ridge Rregressiion"
    ],

    "MAE":[
    mae,
    fe_mae,
    ridge_mae
    ],

    "RMSE":[
     rmse , 
     fe_rmse, 
     ridge_rmse   
    ],

    "R²": [
        r2,
        fe_r2,
        ridge_r2
    ]

})

comparison
```
